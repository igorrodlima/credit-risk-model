## Importing data

Here, I'm importing the data into data\raw.


```python
from pathlib import Path
import subprocess
import zipfile

DATA_DIR = Path("../data/raw")
DATA_DIR.mkdir(parents=True, exist_ok=True)

zip_path = DATA_DIR / "home-credit-default-risk.zip"

# Download from Kaggle
subprocess.run([
    "kaggle",
    "competitions",
    "download",
    "-c",
    "home-credit-default-risk",
    "-p",
    str(DATA_DIR)
], check=True)

# Extract
with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(DATA_DIR)

print(f"Dataset extracted to: {DATA_DIR.resolve()}")
```

I'm converting our CSV files to Parquet for better performance.

```python
import pandas as pd
from pathlib import Path

raw = Path("../data/raw")
proc = Path("../data/processed")
proc.mkdir(parents=True, exist_ok=True)

SKIP = {"HomeCredit_columns_description.csv", "sample_submission.csv"}

assert raw.exists(), f"Pasta não encontrada: {raw.resolve()}"

for f in sorted(raw.glob("*.csv")):
    if f.name in SKIP:
        continue
    out = proc / f"{f.stem}.parquet"
    if out.exists():
        print(f"{f.name}: already converted, skipping")
        continue
    df = pd.read_csv(f)
    df.to_parquet(out, index=False)  
    print(f"File {f.name} converted: {df.shape}")
```
